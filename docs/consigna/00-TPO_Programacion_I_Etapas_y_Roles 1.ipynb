### Estructuras de datos iniciales

Indicar qué estructuras se utilizarán y por qué.

- **Lista:** `carpas`, `clientes` y `reservas` son listas de listas. Cada registro es una lista porque necesita modificarse (cambiar estado, renumerar) y en la Etapa 1 solo contamos con listas como estructura mutable. Se usan para búsquedas lineales y ordenamientos.
- **Matriz:** no se usa en la Etapa 1. Como opción para la Etapa 2, una matriz `ocupacion[carpa][dia]` (0 libre / 1 reservada / 2 alquilada) simplificaría las consultas por quincena y las estadísticas.
- **Strings:** fechas en formato `"dd/mm/aaaa"`, nombre del cliente (normalización y búsqueda parcial), teléfono y `estado` de la reserva.
- **Otras:** ninguna por ahora. En la Etapa 2 se evaluará pasar `clientes` a un diccionario indexado por DNI y `carpa` a una tupla `(num_carpa, activa)`.
